Why Random Tensors?
Randome tensors are imp bcoz the way neural networks learn is they start with tensors full of random. umbers. Then adjust those numbers to better represent the data


```python
matrix_a = torch.randn(3,4)
print(matrix_a)
```

```python
matrix_b = torch.randn(4,2)
print(matrix_b)
```

```python
print("Matrix A Shape:", matrix_a.shape)
print("Matrix B Shape:", matrix_b.shape)
```

```python
# 2. Perform Matrix Multiplication
# In PyTorch, '@' is the operator for matrix multiplication
result = matrix_a @ matrix_b
```

```python
print("\nResulting Matrix Shape:",result.shape)
```

```python
print("\nResulting Data:\n",result)
```

```python
print("\nDevice being used:", result.device)
```

```python
result.ndim
```

```python
# Create a random tensor with similar shape to an image tensor
random_image_size_tensor = torch.rand(size=(224, 224, 3))  # Height x Width x Channels
print(random_image_size_tensor.shape)
print(random_image_size_tensor.ndim)
```

Tensor datatypes
**Note:** Tensor datatype is one of the 3 big errors with PyTorch deep learning
1. Tensor is not right datatype
2.Tensor not right shape
3.Tensor not on the right device

```python
#Float 32 tensor
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                                dtype=None,
                                device=None,
                                requires_grad=False)
print(float_32_tensor.dtype)
print(float_32_tensor)
```

```python
#Convert float 32 to 16
float_16_tensor = float_32_tensor.type(torch.float16)
print(float_16_tensor.dtype)
print(float_16_tensor)
```
